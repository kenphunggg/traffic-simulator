```python
%matplotlib inline 
import matplotlib.pyplot as plt
import numpy as np
import csv
import seaborn as sns

raw_data = 'main.csv'

def plotKDF(raw_data, label, filter):
  treated_data = []
  with open (f"/home/thai/ken-knative/ken/traffic-simulator/result_file/lamda_5/summary/{raw_data}", mode = 'r') as csv_file:
    csv_reader = csv.reader(csv_file, delimiter=',')

    for row in csv_reader:
      if row[2] == filter:
        treated_data.append(float(row[1]))
  sns.kdeplot(treated_data, fill=True, label=label)

# Main
# Init graph
plt.xlabel('Response Time (ms)') 
plt.ylabel('CDF') 
plt.title('CDF of Response Time')

# Plotting graph
plotKDF(raw_data, "Edge", "Konnichiwa from hello-edge-node-e1zri in edge-node")
plotKDF(raw_data, "Cloud", "Konnichiwa from hello-cloud-node-1h7rl in cloud-node")

# Final setup
plt.xlim(0, 1000)
plt.xlim(0, 1)
plt.legend()
plt.show()
     
```
